# 1. What are the key concepts of object-oriented programming (OOPS)?

Object-oriented programming (OOP) is based on several key concepts that help in organizing and structuring software.
Classes and Objects:

1. Class:** A blueprint for creating objects. It defines properties (attributes) and behaviors (methods) that the objects created from the class will have.
2.  Object**: An instance of a class. It represents a specific entity with the properties and behaviors defined by its class.
3.   Encapsulation:**
Bundles the data (attributes) and methods (functions) that operate on the data into a single unit or class. It also restricts direct access to some of the object’s components, which is known as data hiding.
4.Abstraction:**
The concept of simplifying complex reality by modeling classes based on the essential properties and behaviors an object should have, while hiding unnecessary details.
5.  A mechanism that allows a new class (subclass or derived class) to inherit attributes and methods from an existing class (superclass or base class). This promotes code reusability and establishes a hierarchical relationship between classes.
6. Polymorphism:**
The ability for different classes to be treated as instances of the same class through a common interface. This can be achieved through method overriding (where a subclass provides a specific implementation of a method already defined in its superclass) and method overloading (where two or more methods in the same scope have the same name but different parameters).
7.   Composition:
A design principle where a class is composed of one or more objects from other classes, establishing a "has-a" relationship. This allows for creating complex types by combining simpler ones.
These concepts work together to promote modularity, maintainability, and scalability in software design.

# 3. Explain the difference between instance methods and class methods. Provide an example of each.

In Python, instance methods and class methods are two types of methods that serve different purposes within a class.
Instance Methods
Definition: Instance methods are functions defined within a class that operate on instances of that class. They take self as the first parameter, which refers to the specific object (instance) calling the method.
Usage: Instance methods can access and modify the object's attributes.
e.g.

```python
class Dog:
    def __init__(self, name):
        self.name = name

    def bark(self):
        return f"{self.name} says woof!"

# Example usage:
my_dog = Dog("Buddy")
print(my_dog.bark())
```

Class Methods
Definition: Class methods are functions defined within a class that operate on the class itself rather than on instances of the class. They take cls as the first parameter, which refers to the class itself. Class methods are marked with the @classmethod decorator.
Usage: Class methods can be used for factory methods or to access or modify class-level attributes.

```python
class Dog:
    species = "Canis lupus familiaris"  # Class attribute

    def __init__(self, name):
        self.name = name

    @classmethod
    def get_species(cls):
        return cls.species

# Example usage:
print(Dog.get_species())
```


* Instance methods operate on an instance and can access instance attributes.
* Class methods operate on the class itself and can access class attributes.
These distinctions help in organizing code and maintaining proper access to attributes and methods in object-oriented programming.

# 4.How does Python implement method overloading? Give an example.

In Python, method overloading as seen in some other languages (like Java or C++) is not natively supported. However, we can achieve similar functionality using default arguments or by handling different types of arguments within a single method.

Example of Method Overloading using Default Arguments

```python
class MathOperations:
    def add(self, a, b=0):
        return a + b

# Example usage:
math_op = MathOperations()

# Calling with two arguments
result1 = math_op.add(5, 3)
print(result1)

# Calling with one argument
result2 = math_op.add(5)
print(result2)
```

Example of Method Overloading by Handling Types
Another approach is to check the types of the arguments within the method:

```python
class Calculator:
    def multiply(self, a, b):
        if isinstance(a, str) and isinstance(b, int):
            return a * b  # String multiplication
        elif isinstance(a, int) and isinstance(b, int):
            return a * b  # Integer multiplication
        else:
            raise ValueError("Invalid types for multiplication")

# Example usage:
calc = Calculator()

# String multiplication
result1 = calc.multiply("Hello ", 3)
print(result1)

# Integer multiplication
result2 = calc.multiply(4, 5)
print(result2)
```

While Python doesn't have traditional method overloading, we can achieve similar results by using default arguments or by checking the types of arguments within a single method. This provides flexibility in how methods can be called and allows for different behaviors based on input types.

# 5.  What are the three types of access modifiers in Python? How are they denoted?

In Python, access modifiers are used to control the visibility and accessibility of class attributes and methods. The three main types of access modifiers are:

1. Public:
Definition: Public members (attributes and methods) are accessible from outside the class. By default, all members are public in Python.
Denotation: No special notation is used.

example.

```python
class MyClass:
    def __init__(self):
        self.public_attr = "I am public"

obj = MyClass()
print(obj.public_attr)
```

2. Protected:

Definition: Protected members are intended for internal use within the class and its subclasses. They are not meant to be accessed directly from outside the class. However, they can still be accessed if necessary.
Denotation: Denoted by a single underscore (_) prefix.


```python
class MyClass:
    def __init__(self):
        self._protected_attr = "I am protected"

obj = MyClass()
print(obj._protected_attr)
```

3.Private:

Definition: Private members are intended for internal use only and cannot be accessed directly from outside the class. They are meant to be hidden from subclasses and external access.
Denotation: Denoted by a double underscore (__) prefix.

```python
class MyClass:
    def __init__(self):
        self.__private_attr = "I am private"

    def get_private_attr(self):
        return self.__private_attr

obj = MyClass()
# print(obj.__private_attr)  # This would raise an AttributeError
print(obj.get_private_attr())
```



*   Public: Accessible everywhere (no prefix).
*   Protected: Accessible within the class and
*   subclasses (single underscore prefix).
Private: Accessible only within the class (double underscore prefix).



# 6.Describe the five types of inheritance in Python. Provide a simple example of multiple inheritance.

In Python, inheritance allows a class (derived class) to inherit attributes and methods from another class (base class). There are five types of inheritance:

1. Single Inheritance
In single inheritance, a derived class inherits from only one base class.

```python
class Animal:
    def speak(self):
        return "Animal speaks"

class Dog(Animal):
    def bark(self):
        return "Dog barks"

# Usage
dog = Dog()
print(dog.speak())
```

2. Multiple Inheritance
In multiple inheritance, a derived class can inherit from more than one base class.

```python
class Canine:
    def bark(self):
        return "Bark"

class Pet:
    def play(self):
        return "Playing"

class Dog(Canine, Pet):
    def wag_tail(self):
        return "Wagging tail"

# Usage
dog = Dog()
print(dog.bark())
print(dog.play())
print(dog.wag_tail())
```

3. Multilevel Inheritance
In multilevel inheritance, a class is derived from another derived class.

```python
class Animal:
    def speak(self):
        return "Animal speaks"

class Dog(Animal):
    def bark(self):
        return "Dog barks"

class Puppy(Dog):
    def whine(self):
        return "Puppy whines"

# Usage
puppy = Puppy()
print(puppy.speak())
print(puppy.bark())
print(puppy.whine())
```

4. Hierarchical Inheritance
In hierarchical inheritance, multiple derived classes inherit from the same base class.

```python
class Animal:
    def speak(self):
        return "Animal speaks"

class Dog(Animal):
    def bark(self):
        return "Dog barks"

class Cat(Animal):
    def meow(self):
        return "Cat meows"

# Usage
dog = Dog()
cat = Cat()
print(dog.speak())
print(cat.speak())
```

5. Hybrid Inheritance
Hybrid inheritance is a combination of two or more types of inheritance. This can include multiple and hierarchical inheritance, but it can be complex and may lead to ambiguity.

```python
class Base:
    def method(self):
        return "Method from Base"

class A(Base):
    def method_a(self):
        return "Method from A"

class B(Base):
    def method_b(self):
        return "Method from B"

class C(A, B):
    def method_c(self):
        return "Method from C"

# Usage
obj = C()
print(obj.method())
print(obj.method_a())
print(obj.method_b())
print(obj.method_c())
```


*  Single Inheritance: One base class.
*   Multiple Inheritance: Multiple base classes.
*   Multilevel Inheritance: Inheriting from a derived class.
*   Hierarchical Inheritance: Multiple classes from one base class.
*  Hybrid Inheritance: Combination of multiple types.


# 7.What is the Method Resolution Order (MRO) in Python? How can you retrieve it programmatically?

Method Resolution Order (MRO) in Python is the order in which classes are looked up when searching for a method or attribute in a hierarchy of classes. This is particularly important in the context of multiple inheritance, where a method could be inherited from multiple classes, and MRO determines which method gets called.
MRO in Python
Python uses an algorithm called C3 linearization (or C3 superclass linearization) to determine the MRO. The MRO ensures a consistent order for method resolution while maintaining the following properties:

A class always appears before its parents.
If a class appears in the MRO of its parents, it should not appear again later in the order.


Retrieving MRO Programmatically
You can retrieve the MRO of a class using the mro() method or the __mro__ attribute.

```python
class A:
    pass

class B(A):
    pass

class C(A):
    pass

class D(B, C):
    pass

# Retrieve MRO
print(D.mro())
print(D.__mro__)
```



*  D is derived from both B and C, which in turn both inherit from A.
*   The MRO for D is [D, B, C, A, object], meaning that when looking for a method in an instance of D, Python will first check D, then B, then C, followed by A, and finally the base object class

MRO is a crucial aspect of Python's method resolution strategy, especially in multiple inheritance scenarios. You can easily retrieve it using either the mro() method or the __mro__ attribute. This helps understand the order in which classes are searched for methods and attributes, ensuring predictable behavior in complex class hierarchies.

## 8. Create an abstract base class Shape with an abstract method area(). Then create two subclasses# Circle and Rectangle that implement the area() method.

To create an abstract base class Shape with an abstract method area(), we can use the abc module in Python. Shape class and its subclasses Circle and Rectangle:

```python
from abc import ABC, abstractmethod
import math

class Shape(ABC):
    @abstractmethod
    def area(self):
        pass

class Circle(Shape):
    def __init__(self, radius):
        self.radius = radius

    def area(self):
        return math.pi * (self.radius ** 2)

class Rectangle(Shape):
    def __init__(self, width, height):
        self.width = width
        self.height = height

    def area(self):
        return self.width * self.height

# Example usage
circle = Circle(5)
rectangle = Rectangle(4, 6)

print(f"Area of the circle: {circle.area():.2f}")
print(f"Area of the rectangle: {rectangle.area():.2f}")
```

Abstract Base Class:
*   The Shape class is defined as an abstract base
*   The area() method is declared as an abstract method using the @abstractmethod decorator, meaning that any subclass must implement this method.


Subclasses:
*  Circle: Implements the area() method to calculate the area of a circle using the formula  πr 2
* Rectangle: Implements the area() method to calculate the area of a rectangle using the formula width×height

Example Usage:

Instances of Circle and Rectangle are created, and their areas are calculated and printed.

# 9. Demonstrate polymorphism by creating a function that can work with different shape objects to calculate  and print their areas.

Polymorphism allows objects of different classes to be treated as objects of a common superclass. In this case, we can create a function that takes shape objects and calculates their areas, demonstrating polymorphism with the Shape, Circle, and Rectangle classes.

```python
from abc import ABC, abstractmethod
import math

class Shape(ABC):
    @abstractmethod
    def area(self):
        pass

class Circle(Shape):
    def __init__(self, radius):
        self.radius = radius

    def area(self):
        return math.pi * (self.radius ** 2)

class Rectangle(Shape):
    def __init__(self, width, height):
        self.width = width
        self.height = height

    def area(self):
        return self.width * self.height

def print_area(shape: Shape):
    print(f"Area: {shape.area():.2f}")

# Example usage
circle = Circle(5)
rectangle = Rectangle(4, 6)

print_area(circle)
print_area(rectangle)
```

1. Abstract Base Class and Subclasses:We define the Shape class with an abstract method area(), along with the Circle and Rectangle classes that implement this method.

2. Polymorphic Function:The print_area() function takes a parameter shape, which is expected to be an instance of the Shape class or its subclasses. It calls the area() method on the passed shape object and prints the area.

3.Example Usage: Instances of Circle and Rectangle are created, and the print_area() function is called with each object. Despite the objects being of different types, the function works seamlessly due to polymorphism.

This demonstrates how polymorphism allows us to interact with different shapes in a unified way, promoting code reusability and flexibility.

# 10. Implement encapsulation in a BankAccount class with private attributes for balance and account_number. Include methods for deposit, withdrawal, and balance inquiry.

Implementation of the BankAccount Class

```python
class BankAccount:
    def __init__(self, account_number, initial_balance=0):
        self.__account_number = account_number  # Private attribute
        self.__balance = initial_balance          # Private attribute

    def deposit(self, amount):
        if amount > 0:
            self.__balance += amount
            print(f"Deposited: ${amount:.2f}")
        else:
            print("Deposit amount must be positive.")

    def withdraw(self, amount):
        if 0 < amount <= self.__balance:
            self.__balance -= amount
            print(f"Withdrew: ${amount:.2f}")
        elif amount > self.__balance:
            print("Insufficient funds.")
        else:
            print("Withdrawal amount must be positive.")

    def get_balance(self):
        return self.__balance

    def get_account_number(self):
        return self.__account_number

# Example usage
account = BankAccount("123456789", 1000)

print(f"Account Number: {account.get_account_number()}")
print(f"Initial Balance: ${account.get_balance():.2f}")

account.deposit(500)               # Depositing money
print(f"Balance after deposit: ${account.get_balance():.2f}")

account.withdraw(200)              # Withdrawing money
print(f"Balance after withdrawal: ${account.get_balance():.2f}")

account.withdraw(1500)             # Attempting to withdraw more than the balance
```


1.   1.Private Attributes:The __account_number and __balance attributes are marked as private by using double underscores (__). This means they cannot be accessed directly from outside the class.
2.   Methods:
*   deposit(amount): Increases the balance by the specified amount if it's positive.
*   withdraw(amount): Decreases the balance by the specified amount if sufficient funds are available and the amount is positive.
*   get_balance(): Returns the current balance. This method allows controlled access to the private balance.
*   get_account_number(): Returns the account number, providing controlled access to the private account number.
3.  Example Usage:

Example Class

```python
class Vector:
    def __init__(self, x, y):
        self.x = x
        self.y = y

    def __str__(self):
        return f"Vector({self.x}, {self.y})"

    def __add__(self, other):
        if isinstance(other, Vector):
            return Vector(self.x + other.x, self.y + other.y)
        return NotImplemented

# Example usage
v1 = Vector(2, 3)
v2 = Vector(5, 7)

# Using __str__ method
print(v1)
print(v2)

# Using __add__ method
v3 = v1 + v2
print(v3)
```

# 12. Create a decorator that measures and prints the execution time of a function.

```python
import time

def timing_decorator(func):
    def wrapper(*args, **kwargs):
        start_time = time.time()  # Record start time
        result = func(*args, **kwargs)  # Call the actual function
        end_time = time.time()  # Record end time
        execution_time = end_time - start_time  # Calculate execution time
        print(f"Execution time of {func.__name__}: {execution_time:.4f} seconds")
        return result
    return wrapper

# Example usage of the decorator
@timing_decorator
def slow_function(seconds):
    time.sleep(seconds)  # Simulate a slow operation
    return "Finished"

# Call the decorated function
result = slow_function(2)  # This will take approximately 2 seconds
print(result)
```

# 13. Explain the concept of the Diamond Problem in multiple inheritance. How does Python resolve it?

```python
    A
   / \
  B   C
   \ /
    D
```

*   Class A is the base class.
*  Classes B and C inherit from A.
*   Class D inherits from both B and C

If both B and C override a method from A, when D calls this method, it's unclear whether it should inherit the method from B or from C.

How Python Resolves the Diamond Problem
Python uses the C3 Linearization (or C3 superclass linearization) algorithm to resolve the Diamond Problem. This algorithm provides a consistent method resolution order (MRO) to determine the order in which classes are searched for methods and attributes.

Steps in C3 Linearization


1.   Order of Classes: The order in which classes are defined affects the MRO. For class D, the MRO is determined by the order of classes in the inheritance.
2.  Linearization Process: The MRO is computed by creating a list that includes:
*   The class itself (D).
*   The linearized order of the parent classes, starting from the leftmost parent (first) to the rightmost parent (last).
*   The MRO of the base classes, ensuring that each base class appears before its children.
Consistency: The algorithm ensures that a class always appears before its parents in the order and avoids ambiguity.










```python
class A:
    def greet(self):
        return "Hello from A"

class B(A):
    def greet(self):
        return "Hello from B"

class C(A):
    def greet(self):
        return "Hello from C"

class D(B, C):
    pass

# Determine the MRO of class D
print(D.mro())
# Create an instance of D
d = D()
print(d.greet())
```

1.  Class Hierarchy:
*   Classes B and C both inherit from A, and D inherits from both B and C.
2.   Method Resolution Order (MRO):
*   Calling D.mro() shows the order in which Python will search for methods: [D, B, C, A, object]
*  This means that if D calls greet(), it will first look in B, then in C, and finally in A.
3. Result:
The method greet() from B is called, resulting in the output "Hello from B"









# 14.Write a class method that keeps track of the number of instances created from a class.



To create a class method that keeps track of the number of instances created from a class, we  can use a class variable to maintain a count. Each time an instance of the class is created, you can increment this counter in the __init__ method

```python
class InstanceCounter:
    instance_count = 0  # Class variable to keep track of instances

    def __init__(self):
        InstanceCounter.instance_count += 1  # Increment count on instance creation

    @classmethod
    def get_instance_count(cls):
        return cls.instance_count  # Return the current instance count

# Example usage
obj1 = InstanceCounter()
obj2 = InstanceCounter()
obj3 = InstanceCounter()

print(f"Number of instances created: {InstanceCounter.get_instance_count()}")
```



1.   Class Variable
*   instance_count is defined as a class variable. This variable is shared among all instances of the class.
2.  Constructor (__init__):
*   The constructor increments the instance_count every time a new instance of InstanceCounter is created.
3.   Class Method
*   get_instance_count() is a class method (decorated with @classmethod) that returns the current count of instances created. It uses cls to refer to the class itself.
4.   Example Usage:
*   Three instances of InstanceCounter are created, and the total count of instances is printed using the class method.








# 15.Implement a static method in a class that checks if a given year is a leap year.

1.   A year is a leap year if it is divisible by 4.
2.   However, if the year is divisible by 100, it is not a leap year, unless it is also divisible by 400.



```python
class YearUtils:
    @staticmethod
    def is_leap_year(year):
        """Check if the given year is a leap year."""
        if (year % 4 == 0 and year % 100 != 0) or (year % 400 == 0):
            return True
        return False

# Example usage
year = 2024
if YearUtils.is_leap_year(year):
    print(f"{year} is a leap year.")
else:
    print(f"{year} is not a leap year.")

year = 1900
if YearUtils.is_leap_year(year):
    print(f"{year} is a leap year.")
else:
    print(f"{year} is not a leap year.")
```

Explanation

1.   Static Method:
* The method is_leap_year(year) is defined as a static method using the @staticmethod decorator. This means it does not require an instance of the class to be called and can be accessed directly through the class.
2.  Leap Year Logic:
*   The method checks if the given year meets the criteria for being a leap year and returns True or False accordingly.
3.   Example Usage:
*  The method is called with different years to
check if they are leap years, and the results are printed.
